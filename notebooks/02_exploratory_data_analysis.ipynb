```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

# Set visualization style
sns.set_style("whitegrid")
sns.set_palette("husl")
plt.rcParams['figure.figsize'] = (14, 6)
plt.rcParams['font.size'] = 10

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
pd.set_option('display.float_format', '{:.4f}'.format)

print("✓ All libraries imported successfully")
print("✓ Visualization environment configured")
```

```python
# Configuration
DATA_PATH = '../data/processed/preprocessed_data.csv'
TRAIN_PATH = '../data/processed/train_data.csv'
TEST_PATH = '../data/processed/test_data.csv'

# Feature categories
NUMERIC_FEATURES = ['revenue', 'dti_n', 'loan_amnt', 'fico_n', 'experience_c']
CATEGORICAL_FEATURES = ['emp_length', 'purpose', 'home_ownership_n', 'addr_state']
TARGET_COLUMN = 'Default'

print("✓ Configuration loaded")
```

---

## Section 1: Load and Overview Processed Data

```python
print("="*80)
print("LOADING PROCESSED DATA")
print("="*80)

# Load data
df = pd.read_csv(DATA_PATH)
train_df = pd.read_csv(TRAIN_PATH)
test_df = pd.read_csv(TEST_PATH)

print(f"\n✓ Data loaded successfully!")
print(f"\n📊 DATASET OVERVIEW:")
print(f"  • Full dataset shape: {df.shape[0]:,} rows × {df.shape[1]} columns")
print(f"  • Training set shape: {train_df.shape[0]:,} rows × {train_df.shape[1]} columns")
print(f"  • Test set shape: {test_df.shape[0]:,} rows × {test_df.shape[1]} columns")

print(f"\n📋 COLUMNS:")
print(f"  {df.columns.tolist()}")

print(f"\n📈 DATA TYPES:")
print(df.dtypes)

print(f"\n📊 DATASET INFO:")
print(df.info())

print(f"\n🎯 TARGET VARIABLE DISTRIBUTION:")
print(f"  Class 0 (Fully Paid): {(df[TARGET_COLUMN] == 0).sum():,} ({(df[TARGET_COLUMN] == 0).sum()/len(df)*100:.2f}%)")
print(f"  Class 1 (Default): {(df[TARGET_COLUMN] == 1).sum():,} ({(df[TARGET_COLUMN] == 1).sum()/len(df)*100:.2f}%)")

print(f"\n✓ Data loaded and ready for EDA")
```

```python
print("\n📋 FIRST FEW ROWS:")
print(df.head(10))

print("\n📊 STATISTICAL SUMMARY:")
print(df.describe())
```

---

## Section 2: Univariate Analysis (Feature Distributions)

```python
print("\n" + "="*80)
print("UNIVARIATE ANALYSIS: FEATURE DISTRIBUTIONS")
print("="*80)

# Numeric features distribution
numeric_cols = [col for col in df.columns if col != TARGET_COLUMN and df[col].dtype in [np.number]]

print(f"\n📊 NUMERIC FEATURES: {numeric_cols}")
print(f"\nDistribution Statistics:")
for col in numeric_cols:
    print(f"\n  {col}:")
    print(f"    Mean: {df[col].mean():.4f} | Median: {df[col].median():.4f}")
    print(f"    Std Dev: {df[col].std():.4f} | Min: {df[col].min():.4f} | Max: {df[col].max():.4f}")
    print(f"    Skewness: {stats.skew(df[col]):.4f} | Kurtosis: {stats.kurtosis(df[col]):.4f}")

# Visualize distributions
fig, axes = plt.subplots(3, 3, figsize=(16, 12))
axes = axes.ravel()

for idx, col in enumerate(numeric_cols):
    axes[idx].hist(df[col], bins=50, color='steelblue', edgecolor='black', alpha=0.7)
    axes[idx].set_title(f'{col} Distribution', fontweight='bold')
    axes[idx].set_xlabel(col)
    axes[idx].set_ylabel('Frequency')
    axes[idx].grid(alpha=0.3)
    
    # Add statistics text
    stats_text = f"μ={df[col].mean():.2f}\nσ={df[col].std():.2f}"
    axes[idx].text(0.98, 0.97, stats_text, transform=axes[idx].transAxes,
                   fontsize=9, verticalalignment='top', horizontalalignment='right',
                   bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

# Hide unused subplots
for idx in range(len(numeric_cols), len(axes)):
    axes[idx].axis('off')

plt.tight_layout()
plt.show()

print("\n✓ Univariate analysis complete")
```

```python
# Box plots for outlier detection
fig, axes = plt.subplots(1, len(numeric_cols), figsize=(16, 4))

for idx, col in enumerate(numeric_cols):
    axes[idx].boxplot(df[col], vert=True)
    axes[idx].set_title(f'{col} Boxplot', fontweight='bold')
    axes[idx].set_ylabel('Value')
    axes[idx].grid(alpha=0.3, axis='y')

plt.tight_layout()
plt.show()

print("✓ Box plots for outlier visualization complete")
```

---

## Section 3: Bivariate Analysis (Features vs Target)

```python
print("\n" + "="*80)
print("BIVARIATE ANALYSIS: FEATURES vs TARGET")
print("="*80)

# Compare feature distributions by target class
fig, axes = plt.subplots(3, 3, figsize=(16, 12))
axes = axes.ravel()

for idx, col in enumerate(numeric_cols):
    # Separate data by class
    class_0 = df[df[TARGET_COLUMN] == 0][col]
    class_1 = df[df[TARGET_COLUMN] == 1][col]
    
    # Plot distributions
    axes[idx].hist(class_0, bins=30, alpha=0.6, label='Fully Paid (0)', color='green', edgecolor='black')
    axes[idx].hist(class_1, bins=30, alpha=0.6, label='Default (1)', color='red', edgecolor='black')
    axes[idx].set_title(f'{col} Distribution by Target', fontweight='bold')
    axes[idx].set_xlabel(col)
    axes[idx].set_ylabel('Frequency')
    axes[idx].legend()
    axes[idx].grid(alpha=0.3)
    
    # Calculate statistics
    mean_0 = class_0.mean()
    mean_1 = class_1.mean()
    t_stat, p_value = stats.ttest_ind(class_0, class_1)
    
    print(f"\n  {col}:")
    print(f"    Mean (Default=0): {mean_0:.4f}")
    print(f"    Mean (Default=1): {mean_1:.4f}")
    print(f"    T-test p-value: {p_value:.6f} {'✓ Significant' if p_value < 0.05 else '✗ Not Significant'}")

# Hide unused subplots
for idx in range(len(numeric_cols), len(axes)):
    axes[idx].axis('off')

plt.tight_layout()
plt.show()

print("\n✓ Bivariate analysis complete")
```
